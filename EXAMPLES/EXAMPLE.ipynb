### 1.2 Searching Sentinel-2 image products based on the AOI
This section allows to define some parameters for searching Sentinel-2 products. If the user already has the ID of the product, they can skip this and the following cell


```python
# Copernicus Data Space Catalogue OData endpoint
catalogue_odata_url = "https://catalogue.dataspace.copernicus.eu/odata/v1"
collection_name = "SENTINEL-2"
product_type = "S2MSI2A"

# USER DEFINED PARAMETERS
# Define the maximum cloud coverage
max_cloud_cover = 30

#Define the period
search_period_start = "2025-09-15T00:00:00.000Z"
search_period_end = "2025-09-30T00:00:00.000Z"
# END OF USER DEFINED PARAMETERS

aoi = geometry.box(*bbox).wkt
```

```python
# Copernicus Data Space Catalogue is queried based on AOI, period etc
search_query = f"{catalogue_odata_url}/Products?$filter=Collection/Name eq '{collection_name}' and Attributes/OData.CSC.StringAttribute/any(att:att/Name eq 'productType' and att/OData.CSC.StringAttribute/Value eq '{product_type}') and OData.CSC.Intersects(area=geography'SRID=4326;{aoi}') and ContentDate/Start gt {search_period_start} and ContentDate/Start lt {search_period_end} and Attributes/OData.CSC.DoubleAttribute/any(att:att/Name eq 'cloudCover' and att/OData.CSC.DoubleAttribute/Value le {max_cloud_cover})"
response = requests.get(search_query).json()
result = pd.DataFrame.from_dict(response["value"])
result.head(5)
```

### 1.3 Select the image you want to download from the list above
Choose the index

```python
# USER DEFINED PARAMETERS
print("Please select an index based on the results provided above")
Index_image = int(input("Input index: "))
Index_range = len(result)

while Index_image < 0 or Index_image >= Index_range:
    print("Invalid index. Please select a valid index in the range 0 -", Index_range - 1)
    Index_image = int(input("Index: "))

print("Selected index:", Index_image)
```

### 1.4 Set OAuth Client Configuration with Copernicus Dataspace Ecosystem

Before proceding with the download, the user should create an account on Copernicus Data Space and get OAuth client, a guide can be found at this [link](https://documentation.dataspace.copernicus.eu/APIs/SentinelHub/Overview/Authentication.html)

```python
# Create Data Directory
os.makedirs("./DATA", exist_ok=True)
path=f"./DATA/"
# Define variables
Image_Id = result.iloc[Index_image]["Id"]
Image_name = result.iloc[Index_image]["Name"]
PRODUCT_URL = f"https://download.dataspace.copernicus.eu/odata/v1/Products({Image_Id})/$value"
OUTPUT_PATH = path + f"{Image_name}.zip"
# Name of your credentials file with OAuth client id and secret
CREDENTIALS_FILE = "My_credentials.json"
```

```python
# Check if the file exists in the current directory
if os.path.exists(CREDENTIALS_FILE):
    print(f"File '{CREDENTIALS_FILE}' found. Loading credentials...")
    try:
        with open(CREDENTIALS_FILE, 'r') as f:
            credentials = json.load(f)
            client_id = credentials.get("username")
            client_secret = credentials.get("password")
        print(" Credentials loaded successfully from file.")
    except Exception as e:
        print(f" Error reading the file: {e}.Please enter your credentials manually.")
        client_id = input("Enter your OAuth client ID: ")
        client_secret = getpass.getpass("Enter your OAuth client secret: ")
else:
    # If file is missing, ask the user
    print(f"File '{CREDENTIALS_FILE}' not found. Please enter your credentials manually.")
    client_id = input("Enter your OAuth client ID for COPERNICUS DATASPACE ECOSYSTEM: ")
    client_secret = getpass.getpass("Enter your OAuth client secret for COPERNICUS DATASPACE ECOSYSTEM: ")      # getpass hides the input for security

print(f"--> Ready to establish a connection to COPERNICUS.")
```

### 1.5 Authentication (get token)

```python
# Get an access token and establish a session
token_url = "https://identity.dataspace.copernicus.eu/auth/realms/CDSE/protocol/openid-connect/token"
token_payload = {
    "grant_type": "client_credentials",
    "client_id": client_id,
    "client_secret": client_secret
}
try:
    # Perform the POST request to get the token
    auth_response = requests.post(token_url, data=token_payload)
    auth_response.raise_for_status() # Stop if credentials are wrong
    # Parse the JSON response
    token_data = auth_response.json()
    access_token = token_data["access_token"]
    expires_in = token_data["expires_in"]
    print(f"SUCCESS: Token response {auth_response.status_code}. Expires in {expires_in} hours.")
except requests.exceptions.RequestException as e:
    print(f" ERROR: Authentication failed {e}")
    print(f"Server Response: {auth_response.text}")
    sys.exit(1) # Stop the script
```

### 1.6 Download the data

```python
# Download the image, this might take few minutes
try:
    # Create a session to manage the connection headers
    session = requests.Session()
    session.headers["Authorization"] = f"Bearer {access_token}"
    response = session.get(PRODUCT_URL, stream=True)
    response.raise_for_status()
    zip_data = response.content
    response.close()
    print(f"Starting download from: {PRODUCT_URL}")

    # Writes the downloaded content to a file
    with open(OUTPUT_PATH, 'wb') as f:
        f.write(zip_data)

    print(f"\nData saved to: '{OUTPUT_PATH}' (Size: {len(zip_data)} bytes).")

except requests.exceptions.RequestException as e:
    print(f"\nAn error occurred during the request: {e}")
```

# 2. Run Zipra Library Example

### 2.1 Band extraction function
This first section is used to produce a GeoTIFF file containing the selected bands from Sentinel 2 .SAFE file.
If no bands are provided, it extracts the bands: B02, B03, B04, B08, B12, SCL by default

### 2.2 Clip AOI
This section defines a clip for the raster file and the raster file clipped

```python
# Draw the geometry to get the clipped file
#first get the crs from the tiff_file
with rasterio.open(tiff_file) as src:
    crs_file = src.crs.to_string()
    bbox = src.bounds

# trasform in epsg 4326 just for the plot

transformer = Transformer.from_crs(crs_file, "EPSG:4326", always_xy=True)
lon_min, lat_min = transformer.transform(bbox[0], bbox[1])
lon_max, lat_max = transformer.transform(bbox[2], bbox[3])

# Map in EPSG:4326
m2 = leafmap.Map(center=((lat_min+lat_max)/2, (lon_min+lon_max)/2), zoom=8, epsg="4326", toolbar_control=False)
# add the rectangle defined before
#   if the rectangle doesn't coincide with the first draw in leafmap
#   it's due to the reprojection in UTM, done in Band_extraction
rect = Rectangle(bounds=((lat_min, lon_min), (lat_max, lon_max)), color="orange", fill_opacity=0.1)
rect.name = "Original BBOX"
m2.add_layer(rect)
m2.add_layer_control()
m2
```

```python
# This section extract the AOI clipped from the map and extracts its bounding box
AOI_clipped = m2.draw_features[-1] #last drawn feature
if not AOI_clipped:
    print("Please, select an area of interest on the map")
else:
    AOI_clipped = shape(AOI_clipped['geometry'])
    bbox_clipped = AOI_clipped.bounds
    print(f"The Shapely geometry clipped: {AOI_clipped}")
    print(f"The bounding box in EPSG:4326 is (min_x, min_y, max_x, max_y): {bbox_clipped}")
```

```python
# Call to the function Clip_AOI 

clipped_tiff = Clip_AOI(tiff_file, AOI_clipped.wkt) 
print("Clipped raster data path:", clipped_tiff)
```

```python
# optional plot to see the raster added to the map
m2.add_raster(clipped_tiff, layer_name="Raster clipped")
m2
```

### 2.3 Calculate Spectral Indices
This section calculates the spectral indices from Sentinel-2 bands and added as new layers to the GeoTIFF file.

Supported indices: NDVI, NBR, NDWI, NDMI, NDBI, SAVI, EVI

```python
# Calculate default indices (NDVI, NBR, NDWI)
tiff_with_indices, calculated = Indices_calculation(clipped_tiff)
print(f"Indices calculated: {calculated}")
print(f"Output file: {tiff_with_indices}")
```

```python
# Optional input to select different indices respect to the default ones

valid_indices = ["NDVI", "NBR", "NDWI", "NDMI", "NDBI", "SAVI", "EVI"]

print("Please select a list of indices to be calculated on the input raster")
index_list = input("Class list (comma separated): ")

# Conversion in a list
index_list = [x.strip().upper() for x in index_list.split(",")]

# Checks that all indeces are valid
while not all(idx in valid_indices for idx in index_list):
    print("Invalid input. Allowed indices are:", ", ".join(valid_indices))
    index_list = input("Class list (comma separated): ")
    index_list = [x.strip().upper() for x in index_list.split(",")]

print("Selected indices:", index_list)

tiff_with_indices, calculated = Indices_calculation(tiff_file, index_list)
```

```python
# Optional print of the index statistics 
with rasterio.open(tiff_with_indices) as src:
    band_names = src.descriptions #gets band names

    for i, band_name in enumerate(band_names, start=1):
        if band_name.upper() in calculated:
            data = src.read(i)
            print(f"{band_name} (Band {i}): "
                  f"Min={np.nanmin(data):.2f}, "
                  f"Max={np.nanmax(data):.2f}, "
                  f"Mean={np.nanmean(data):.2f}")
```

```python
# Optional plot of the indeces (need to be run the cell above for band_names input)
# Palette predefined for each index
colormaps = {
    "NDVI": plt.cm.RdYlGn,
    "NBR": plt.cm.RdYlGn,
    "NDWI": plt.cm.RdBu,
    "NDMI": plt.cm.BrBG,
    "NDBI": plt.cm.PuOr,
    "SAVI": plt.cm.YlGn,
    "EVI": plt.cm.RdYlBu
}

with rasterio.open(tiff_with_indices) as src:

    fig, axes = plt.subplots(1, len(calculated), figsize=(6*len(calculated), 5))
    if len(calculated) == 1:
        axes = [axes]  # handles the case with len = 1 

    for ax, idx in zip(axes, calculated):
        if idx in band_names:
            band_num = band_names.index(idx) + 1 
            data = src.read(band_num)

            cmap = colormaps.get(idx, plt.cm.viridis)  
            cmap.set_bad(color="white")

            im = ax.imshow(data, cmap=cmap, vmin=-1, vmax=1)
            ax.set_title(idx)
            fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
        else:
            ax.set_title(f"{idx} not found")

    plt.tight_layout()
    plt.show()
```

### 2.4 Area calculations
This function asks for some specified classes and calculates both the total area and the classes area for the given GeoTIFF file

```python
# Class input
print("Please insert a list of classes for which you want the area calculation")
print("- Class values range: 0 - 11")

valid = False
class_list = []

while not valid:
    user_input = input("Class list (comma separated): ")
    try:
        # conversion in a list of integers
        class_list = [int(x.strip()) for x in user_input.split(",")]
        
        # checks that all values are in the range
        if all(0 <= x <= 11 for x in class_list):
            valid = True
        else:
            print("Invalid input: values must be between 0 and 11")
    except ValueError:
        print("Invalid input: please insert numbers separated by commas")

print("Valid classes saved:", class_list)
```

```python
# Call to Area_calculation function

SCL_band = band_list.index('SCL') + 1  # Get the band number for SCL
Area_tot, Area_classes = Area_calculation(clipped_tiff, class_list, SCL_band)
print("Total area of specified classes [m^2]:", Area_tot)
for i, class_value in enumerate(class_list):
    print(f"Area of class {class_value} [m^2]: {Area_classes[i]}")
```

### 2.5 Mask tiff
This section asks for some specified classes and returns the original file masked accordingly to the specified classes

```python
# Optional different class input
print("Please insert a list of classes for which you want the area calculation")
print("- Class values range: 0 - 11")

valid = False
class_list = []

while not valid:
    user_input = input("Class list (comma separated): ")
    try:
        # conversion in a list of integers
        class_list = [int(x.strip()) for x in user_input.split(",")]
        
        # checks that all values are in the range
        if all(0 <= x <= 11 for x in class_list):
            valid = True
        else:
            print("Invalid input: values must be between 0 and 11")
    except ValueError:
        print("Invalid input: please insert numbers separated by commas")

print("Valid classes saved:", class_list)
```

```python
# Call to the Mask_tiff function (on tiff file with indices)

masked_tiff = Mask_tiff(tiff_with_indices,class_list,SCL_band)
print("Masked tiff file path:", masked_tiff)
```

```python
# Visualization parameters
V_MIN = 0
V_MAX = 3000

def normalize_fixed(band, v_min, v_max):
    band_float = band.astype(float)
    band_float[band == -9999] = np.nan
    
    # Stretching form v_min to v_max
    clipped = np.clip(band_float, v_min, v_max)
    
    # Stretching form 0 to 1
    normalized = (clipped - v_min) / (v_max - v_min)
    
    return normalized

with rasterio.open(masked_tiff) as src:
    r = src.read(3) #red
    g = src.read(2) #green
    b = src.read(1) #blue

    r_norm = normalize_fixed(r, V_MIN, V_MAX)
    g_norm = normalize_fixed(g, V_MIN, V_MAX)
    b_norm = normalize_fixed(b, V_MIN, V_MAX)

    # Stack RGB
    rgb = np.dstack((r_norm, g_norm, b_norm))

    # Plot
    plt.figure(figsize=(10, 10))
    plt.imshow(rgb) 
    plt.axis("off")
    plt.title(f"RGB Composite (Stretch {V_MIN}-{V_MAX})")
    plt.show()
```

### 2.5 Barplot of raster classes
Gets the classes and respective count from the tiff file to return the corresponding histogram.
It also asks if the user wants as additional output the statistics of the file (max, min, mean)

```python
# if stats true --> returns max,min,mean value
stats = input("The output should contain the statistics of the file? [Y/N]: ")

while stats != "Y" and stats != "N":
    stats = input('Not valid answer, please answer "Y" for Yes and "N" for No: ')

print("Selected stats = ", stats)
if stats == "Y": 
    stats = True
    # get the results for the masked tiff obtained 
    unique_classes, counts, stats_res = Barplot_classes(masked_tiff, SCL_band, stats)
    print(f"Max:  {stats_res['max_value']} (Class {stats_res['max_class']})")
    print(f"Min:  {stats_res['min_value']} (Class {stats_res['min_class']})")
    print(f"Mean: {stats_res['mean_value']} (Class {stats_res['mean_class']})")
else:
    stats = False
    # get the results for the masked tiff obtained
    unique_classes, counts = Barplot_classes(masked_tiff, SCL_band, stats)

# Plot
# Define SCL class labels and corresponding ESA/Sen2Cor colors
scl_classes = {
    0: ("0 - No Data", "#000000"),
    1: ("1 - Saturated/Defective", "#ff0000"),
    2: ("2 - Dark Features/Shadows", "#464646"),
    3: ("3 - Cloud Shadows", "#686868"),
    4: ("4 - Vegetation", "#00ff00"),
    5: ("5 - Bare Soils", "#c2b280"),
    6: ("6 - Water", "#0000ff"),
    7: ("7 - Unclassified", "#ffffff"),
    8: ("8 - Cloud Medium Probability", "#bfbfbf"),
    9: ("9 - Cloud High Probability", "#dcdcdc"),
    10: ("10 - Thin Cirrus", "#a5f3ff"),
    11: ("11 - Snow/Ice", "#00ffff")
}

colors = [scl_classes[c][1] for c in unique_classes]
labels = [scl_classes[c][0] for c in unique_classes]

plt.bar(labels, counts*100/sum(counts), color=colors)
plt.xlabel("Classes")
plt.ylabel("Freq %")
plt.xticks(rotation=45, ha="right")  # rotation 45°
plt.title("Classes histogram")
plt.show()
```
